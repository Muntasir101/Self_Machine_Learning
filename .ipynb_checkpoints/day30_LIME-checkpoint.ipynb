# LIME

```python
# Step 1: Dataset
import numpy as np 

# [study_hours, sleep_hour]

X = np.array([
    [2,5],
    [3,6],
    [4,7],
    [5,3],
    [1,4],
    [8,7],
    [5,5],
    [6,4]
])

## 0 = fail, 1 = pass

y= np.array([0,0,0,1,0,1,1,1])

# step 2: Model train
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier()
model.fit(X,y)

from lime.lime_tabular import LimeTabularExplainer 

explainer = LimeTabularExplainer(
    X, 
    feature_names =['study_hours', 'sleep_hours'],
    class_names =['fail', 'pass'],
    mode = 'classification'
)


# Step 4: sample data prediction 
sample = np.array([5,6])
print("Prediction", model.predict([sample]))

#step 5: LIME Explanations
exp = explainer.explain_instance(
    sample, 
    model.predict_proba,
    num_features = 2
)

for feature, weight in exp.as_list():
    print(feature, weight)
```

# Apply Decision Tree

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# initialize and train
dtc_model = DecisionTreeClassifier(random_state=42)
dtc_model.fit(X,y)

# make prediction
y_pred = dtc_model.predict(X)

print("Decision Tree Prediction on training data:", y_pred)
print("Actual Labels:",y)

#accuracy
accuracy = accuracy_score(y, y_pred)
print(f"Decision Tree Model Accuracy: {accuracy:.2f}")

# Step 4: sample data prediction 
sample = np.array([5,6])
print("Prediction: ", model.predict([sample]))

#step 5: LIME Explanations
exp = explainer.explain_instance(
    sample, 
    model.predict_proba,
    num_features = 2
)
for feature, weight in exp.as_list():
    print(feature, weight)
```
