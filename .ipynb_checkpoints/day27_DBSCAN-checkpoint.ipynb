```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
```

```python
X = np.array([[1,1],[1.2,1.2],[2.1,2.1], [5,1],[6.5,5.5],[8,8], [15,15]])
```

```python
dbscan = DBSCAN(eps=2, min_samples=2)
clusters = dbscan.fit_predict(X)
```

```python
plt.figure(figsize=(8,5))
```

```python
plt.scatter(X[:,0], X[:,1], c=clusters, cmap='viridis', s=100, edgecolors='black') 
labels = ['A', 'B', 'C', 'D', 'E', 'F', 'G']
for i, txt in enumerate(labels):
    plt.annotate(txt, (X[i,0]+0.2, X[i,1]))
plt.title("DBSCAN Visualization")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.show()
```
